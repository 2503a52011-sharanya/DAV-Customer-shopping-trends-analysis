```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("/content/customer_shopping_behavior.csv")
df
df.info()
```

```python
# Which product category has the most purchases?
df['Category'].value_counts()
```

```python
# Which category has the highest total purchase amount?
df.groupby('Category')['Purchase Amount (USD)'].sum().sort_values(ascending=False)
```

```python
# Which season generates the highest sales?
highest_sales = df.groupby('Season')['Purchase Amount (USD)'].sum().sort_values(ascending=False)
highest_sales
```

```python
# Which gender has a higher average purchase amount?
gender_analysis = df.groupby("Gender")["Purchase Amount (USD)"].mean().sort_values(ascending=False)
gender_analysis
```

```python
# Which product categories receive the highest customer ratings
rating = df.groupby("Category")["Review Rating"].mean().sort_values(ascending=False)
rating
```

```python
# Which products generate the highest total purchase amount?
product_sales = (
    df.groupby("Item Purchased")["Purchase Amount (USD)"]
    .sum()
    .sort_values(ascending=False)
)
product_sales
```

```python
# Which purchase frequency has the highest average purchase amount?
purchase_frequency = df.groupby("Frequency of Purchases")["Purchase Amount (USD)"].mean()
purchase_frequency
```

```python
# Creating the age groups colunm
df['Age Group'] = pd.cut(
    df['Age'],
    bins=[0, 18, 25, 35, 45, 55, 100],
    labels=['Under 18', '18–25', '26–35', '36–45', '46–55', 'Above 55']
)
df


```

```python
# Which age group has the highest number of customers?
age_group = df['Age Group'].value_counts().sort_values(ascending=False)
age_group
```

```python
# Does a higher previous purchase count correspond to higher customer spending?
df.groupby('Previous Purchases')['Purchase Amount (USD)'].mean()
```

```python
# Which products are purchased most frequently?
top_products = df["Item Purchased"].value_counts().head(10)
top_products
```
