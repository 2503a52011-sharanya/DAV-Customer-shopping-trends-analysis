```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("/content/customer_shopping_behavior.csv")
df
```

```python
# bar chart
df["Category"].value_counts().plot(kind="bar")

plt.title("Customers by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# pie chart
df["Gender"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Customer Distribution by Gender")
plt.ylabel("")
plt.show()
```

```python
# histoggram
plt.hist(df["Purchase Amount (USD)"], bins=10)

plt.title("Distribution of Purchase Amounts")
plt.xlabel("Purchase Amount (USD)")
plt.ylabel("Frequency")
plt.show()
```

```python
# Heatmap
data = df.select_dtypes(include="number")
correlation = data.corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm")

plt.title("Customer Shopping Correlation Heatmap")
plt.show()
```

```python
# box plot
sns.boxplot(
    data=df,
    x="Category",
    y="Purchase Amount (USD)"
)

plt.title("Purchase Amount by Category")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Count plot
sns.countplot(data=df, x="Payment Method")

plt.title("Customers by Payment Method")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Line chart
average = df.groupby("Season")["Purchase Amount (USD)"].mean()

plt.plot(average.index, average.values, marker="o")

plt.title("Average Purchase Amount by Season")
plt.xlabel("Season")
plt.ylabel("Average Purchase Amount")
plt.show()
```
